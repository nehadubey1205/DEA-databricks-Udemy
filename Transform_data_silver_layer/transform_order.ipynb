deduplicate array

```sql
SELECT json_value.order_id, 
       json_value.payment_method,
       json_value.order_status,
       json_value.items,
       json_value.customer_id,
       array_distinct(json_value.items) as items
FROM gizmobox.silver.json_orders;
```

explored 

```sql
SELECT json_value.order_id, 
       json_value.payment_method,
       json_value.order_status,
       json_value.items,
       json_value.customer_id,
       explode(array_distinct(json_value.items)) as item
FROM gizmobox.silver.json_orders;
```

```sql
CREATE OR REPLACE TEMP VIEW tv_orders_exploded
AS
SELECT json_value.order_id, 
       json_value.payment_method,
       json_value.order_status,
       json_value.items,
       json_value.customer_id,
       explode(array_distinct(json_value.items)) as item

FROM gizmobox.silver.json_orders;
```

```sql
SELECT * FROM tv_orders_exploded;
```

```sql
SELECT order_id,
       payment_method,
       order_status,

       item.item_id,
       item.name
FROM tv_orders_exploded;
```

```sql
CREATE TABLE if NOT EXISTS gizmobox.silver.json_orders
AS 
SELECT order_id,
       payment_method,
       order_status,

       item.item_id,
       item.name
FROM tv_orders_exploded;

```

Monthly order summary

```sql
SELECT * FROM gizmobox.silver.json_orders;
```

```sql
SELECT date_format(transaction_timestap,'yyyy-MM') as order_month,
       customer_id,
       count(distinct order_id) as orders
       sum(quantity) as total,
FROM gizmobox.silver.json_orders;
```
