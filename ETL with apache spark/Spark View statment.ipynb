Register files in unity catalog using view

```sql
CREATE OR REPLACE VIEW gizmobox.bronze.v_customers as 
SELECT *,
 _metadata.file_path AS file_path
FROM json.`/Volumes/gizmobox/landing/operation_data/customers`
```

```sql
SELECT * FROM gizmobox.bronze.v_customers;
```

Temoratry view: this exist till spark is on active: once spark session stop this will remove
Global temporary view: this is active till spark app is live/active means onnce culter destroy or terminated this will also remove

```sql
CREATE OR REPLACE TEMPORARY VIEW tv_customers as 
SELECT *,
 _metadata.file_path AS file_path
FROM json.`/Volumes/gizmobox/landing/operation_data/customers`
```

```sql
SELECT * FROM tv_customers; 
```
