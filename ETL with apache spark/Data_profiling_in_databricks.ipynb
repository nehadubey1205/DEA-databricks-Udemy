Manually

```sql
SELECT count(*) FROM gizmobox.bronze.v_customers;
```

```sql
SELECT count(*), count(customer_id)
FROM gizmobox.bronze.v_customers;
```

```sql
SELECT count(*), count_if(customer_id is null)
FROM gizmobox.bronze.v_customers;
```

```sql
SELECT count(*)
FROM gizmobox.bronze.v_customers
where customer_id is null
```

```sql
SELECT count(*) total_number_of_records
 , count(distinct customer_id) unique_customer_id
FROM gizmobox.bronze.v_customers;
```

```sql
SELECT count(*) total_number_of_records
 , count(distinct customer_id) unique_customer_id
FROM gizmobox.bronze.v_customers
where customer_id is not null;
```
