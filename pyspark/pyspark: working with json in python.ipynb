```python
df = spark.read.format("json").load("/Volumes/gizmobox/landing/operation_data/customers/customers_2024_10.json")
display(df)
```

select metadata

```python
df_with_metadata = df.select("_metadata.file_path","*")
display(df_with_metadata)
```

create table in bronze schema

```python
df_with_metadata.write.format("delta").mode("overwrite").saveAsTable("gizmobox.bronze.py_customer")
```

```python
%sql
SELECT * FROM gizmobox.bronze.py_customer;
```

dataframe writeV2 API

```python
df_with_metadata.writeTo("gizmobox.bronze.py_customer").createOrReplace()
```
