spark: create or replace return function

```sql
CREATE or REPLACE FUNCTION gizmobox.default.get_full_name(firstname STRING, surname STRING)
RETURNS string
RETURN CONCAT(initcap(firstname), ' ', initcap(surname));
```

```sql

SELECT gizmobox.default.get_full_name('john', 'saity');
```

```sql
DESC FUNCTION EXTENDED gizmobox.default.get_full_name;
```

Higher order function
1) transform
2) FIlter
3) exists
4) aggregate

```sql
CREATE or REPLACE TEMPORARY VIEW order_items AS SELECT * FROM 
VALUES
    (1, array('smartphone','laptop','monitor')),
    (2, array('tablet','headphones','speaker')),
    (3, array('keyborad','mouse'))
AS orders(order_id, items);
```

```sql
SELECT * FROM order_items;
```

transform

```sql
SELECT order_id,
      transform(items, x -> Upper(x)) AS upper_items
FROM order_items;
```

Filter

```sql
SELECT order_id,
      filter(items, x -> x LIKE '%smart%') AS smart_items
FROM order_items; 
```

Exist

this give value true or false

```sql
SELECT order_id,
      EXISTS(items, x -> x = 'moniter') AS has_items
FROM order_items;
```

Upper + 10% tax

```sql
SELECT order_id,
    TRANSFORM(CAST(items AS ARRAY<STRUCT<name: STRING, price: DOUBLE>>), x -> named_struct(
        'name', UPPER(x.name),
        'price', ROUND(x.price * 1.10, 2)
    )) AS items_with_tax
FROM order_items;
```
