MAP function

```sql
CREATE OR REPLACE VIEW student_scores AS
SELECT * FROM VALUES
  (1, map('math', 85, 'science', 90, 'english', 78)),
  (2, map('math', 92, 'science', 88, 'english', 95)),
  (3, map('math', 70, 'science', 75, 'english', 80))
AS t(student_id, scores);
```

```sql
SELECT * FROM student_scores;
```

```sql
SELECT student_id,
       scores['math'] AS math_score
FROM student_scores;
```

get keys

```sql
SELECT student_id,
       map_keys(scores) AS subjects
FROM student_scores;
```

get all values 

```sql
SELECT student_id,
       map_values(scores) AS marks
FROM student_scores;
```

keep value >80

```sql
SELECT student_id,
       map_filter(scores, (k, v) -> v > 80) AS filtered_scores
FROM student_scores;
```

Convert map to array

```sql
SELECT student_id,
       map_entries(scores) AS key_value_pairs
FROM student_scores;
```
